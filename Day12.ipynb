```python
#handle missing value

import numpy as np
import pandas as pd
data = {
    "experience":[1,np.nan,3,4,5],
    "salary":[100,200,300,400,np.nan],
    "name":['hello','dhreej','Nikhil','Raj','vinay']
}
df = pd.DataFrame(data)
print(df)

# null value filter | find out

print(df.isnull())

# sum of null
print(df.isnull().sum)


# dropna | drop complete row
df.dropna()
print(df)

# fillna | null -> fill
df.fillna(5)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2,2)

# Pie Chart
x = ["Apple","Banana","Carrot","Mango"]
y = [30,50,70,5]
ax[0,0].pie(y, labels=x)
ax[0,0].set_title("Pie Chart")

# Line Chart
x = [1,2,3,4,5]
y = [100,200,300,400,500]
ax[0,1].plot(x,y)
ax[0,1].set_title("Line Chart")

# Bar Chart
x = ["Apple","Banana","Carrot","Mango"]
y = [30,50,70,5]
ax[1,0].bar(x,y)
ax[1,0].set_title("Bar Chart")

# Line Chart
ax[1,1].plot(["Apple","Banana","Carrot","Mango"], [30,50,70,5])
ax[1,1].set_title("Line Chart")

plt.tight_layout()
plt.show()
```
