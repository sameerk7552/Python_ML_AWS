```python
#handle missing value

import numpy as np
import pandas as pd
data = {
    "experience":[1,np.nan,3,4,5],
    "salary":[100,200,300,400,np.nan],
    "name":['hello','dhreej','Nikhil','Raj','vinay']
}
df = pd.DataFrame(data)
print(df)

# null value filter | find out

print(df.isnull())

# sum of null
print(df.isnull().sum)


# dropna | drop complete row
df.dropna()
print(df)

# fillna | null -> fill
df.fillna(5)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2,2)

# Pie Chart
x = ["Apple","Banana","Carrot","Mango"]
y = [30,50,70,5]
ax[0,0].pie(y, labels=x)
ax[0,0].set_title("Pie Chart")

# Line Chart
x = [1,2,3,4,5]
y = [100,200,300,400,500]
ax[0,1].plot(x,y)
ax[0,1].set_title("Line Chart")

# Bar Chart
x = ["Apple","Banana","Carrot","Mango"]
y = [30,50,70,5]
ax[1,0].bar(x,y)
ax[1,0].set_title("Bar Chart")

# Line Chart
ax[1,1].plot(["Apple","Banana","Carrot","Mango"], [30,50,70,5])
ax[1,1].set_title("Line Chart")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Create 2 rows and 2 columns of subplots
fig, ax = plt.subplots(2, 2, figsize=(8, 6))

# ---------------- Pie Chart ----------------
x = ["apple", "banana", "carrot", "mango"]
y = [30, 50, 70, 5]

ax[0, 0].pie(y, labels=x)
ax[0, 0].set_title("Pie Chart")

# ---------------- Line Chart ----------------
data = {
    "experience": [1, 2, 3, 4, 5],
    "salary": [100, 200, 300, 400, 500]
}

x = data["experience"]
y = data["salary"]

ax[0, 1].plot(x, y)
ax[0, 1].set_title("Line Chart")
ax[0, 1].set_xlabel("Experience")
ax[0, 1].set_ylabel("Salary")

# ---------------- Bar Chart ----------------
x = ["apple", "banana", "carrot", "mango"]
y = [30, 50, 70, 5]

ax[1, 0].bar(x, y)
ax[1, 0].set_title("Bar Chart")

# ---------------- Multiline Chart ----------------
x = ["apple", "banana", "carrot", "mango"]
y1 = [30, 50, 70, 5]
y2 = [20, 40, 60, 10]

ax[1, 1].plot(x, y1, marker="o", label="2024")
ax[1, 1].plot(x, y2, marker="s", label="2025")
ax[1, 1].set_title("Multiline Chart")
ax[1, 1].legend()

# Adjust spacing
plt.tight_layout()

# Display all plots
plt.show()
```
