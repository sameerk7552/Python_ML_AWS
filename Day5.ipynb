```python
class student:
    def __init__(self, name, roll_no):
        self.name = name
        self.roll_no = roll_no

s = student("nikhil", 123)

print(s.name)
```

```python
class data:
    def __init__(self,name,id,age):
        self.name=name
        self.id=id
        self.age=age


    def display(self):
        print("NAME:",self.name)
        print("ID:",self.id)
        print("AGE",self.age)


while True:
    name=input("enter your name:")
    id=int(input("enter your id:"))
    age=int(input("enter your age:"))
    c1=data(name,id,age)
    c1.display()
    choice=input("do you want to continue the entries if yes test 'y' other wise 'n':")
    if choice.lower()!='y':
        break
```
