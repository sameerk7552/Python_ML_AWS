```python
class Student:
  def __init__(self,name,roll_no):
    self.name = name
    self.roll_no = roll_no
  # get
  def getName(self):
    print(self.name)
  # update
  def updateName(self):
    self.name = "hello"
    print(self.name)
  # delete
  def deleteName(self):
    del self.name
    # print(self.name)
  # create
  def createName(self):
    self.full_name = "sameer kumar"
    print(self.full_name)

s = Student("nikhil",123)
s.createName()

```

```python
class Student:
  def __init__(self,name,roll_no):
    print("auto start")
    self.name = name
    self.roll_no = roll_no
  # get
  def updateName(self,name=None): # default value -> none
    if name != None:
      self.name = name
    print(self.name)


inst = Student("nikhil",123)
inst.updateName("rajendra")

inst1 = Student("abhishek",321)
inst1.updateName()
```
