```python
class student:
  name = "Nikhil"
  roll_no = 123

s= student() # s is object
print(s.name)  # name is attribute
```

```python
class student:
    def __init__(self, name, roll_no):
        self.name = name
        self.roll_no = roll_no

s = student("nikhil", 123)

print(s.name)
```

```python
class Student:
  def __init__(this,name,roll_no,age):
    this.name = name
    this.roll_no = roll_no
    this.age = age

# s = Student("nikhil",123,20)
# print(s.name,s.roll_no,s.age)

s = Student(roll_no=123,name='nikhil',age=20)
print(s.name,s.roll_no,s.age)
```

```python
class Student:
  def __init__(this,name,roll_no,age=None):
    this.name = name
    this.roll_no = roll_no
    this.age = age

s = Student("nikhil",123)
print(s.name,s.roll_no,s.age)
```
