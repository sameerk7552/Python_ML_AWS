```python
#filter
import numpy as np
arr = np.array([10,20,3,4,90,200])
arr_f = arr[arr > 20]
print(arr_f)
```

```python
import numpy as np
arr = np.array([10,20,30,6,3,2])
arr_f = arr[arr !=6]
print(arr_f)
```

```python
# example and opr 10 se 20 between

import numpy as np
arr = np.array([10,20,30,15,3,2,18,8])
arr_f = arr[(arr>10) & (arr<20)]
print(arr_f)
```

```python
#Fancy indexing

import numpy as np
arr = np.array([10,20,3,4,90,100])
print(arr)
arr_f = arr[[0,2]]# 0 index value , 2 index value
print(arr_f)
```

```python
# where
import numpy as np
arr = np.array([10,20,3,4,90,100])
print(arr)
arr_w = np.where(arr>9,"True","False")
print(arr_w)
```

```python
# question 10 20 -> *2 | +2
import numpy as np
arr = np. array([10,20,3,4,90,100])
arr_w= np.where((arr>10) & (arr<20),arr*2,arr+2)
print(arr_w)
```

```python
# adding data and removing data

# concatenate using 1D
import numpy as np
arr1 = np.array([10,20,30])
arr2 = np.array([1,2,3])

#concatenate
arr_arr2 = np.concatenate((arr1,arr2))
print(arr_arr2)

# concatenate using 2D
arr1 = np.array([[10,20,30],[40,50,60]])
arr2 = np.array([[1,2,3],[4,5,6]])
arr1_arr2 = np.concatenate((arr1,arr2),axis=1) # for row 0, for column 1 but in case of sorting it  get reversed

print(arr1_arr2)
```

```python
# v stack -> vertically join | row-wise
import numpy as np
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
result = np.vstack((arr1,arr2))
print(result)
```

```python
# h stack -> horizontal join | column wise
import numpy as np
arr1 = np.array([10,20,30])
arr2 = np.array([40,50,60])
result = np.hstack((arr1,arr2))
print(result)
```

```python
# for 2d v stack
import numpy as np
arr1 = np.array([[1,2,3],[4,5,6]])
arr2 = np.array([[4,5,6],[7,8,9]])
result = np.vstack((arr1,arr2))
print(result)

# for h stack
arr1 = np.array([[10,20,30],[40,50,60]])
arr2 = np.array([[40,50,60],[70,80,90]])
result = np.hstack((arr1,arr2))
print(result)
```

```python
# statical function

import numpy as np
arr = np.array([2,4,6,8,10])
print(np.sum(arr)) # sum
print(np.min(arr))#min
print(np.max(arr))# max
print(np.mean(arr))# mean
print(np.median(arr)) #median


# np.sum(a) -> Sum of all elements
# np.mean(a) -> Average = Sum of elements / Number of elements
# np.median(a) -> Middle value after sorting
# np.min(a) -> Smallest value in array
# np.max(a) -> Largest value in array
# np.var(a) mean,difference,square,average | (sum = ddof) -> divide by array length
# np.std(a) -> Standard Deviation = √Variance -> Measures spread of data
# np.prod(a) -> Multiplication of all elements
# np.cumsum(a) -> Cumulative (running) sum
# np.cumprod(a) -> Cumulative (running) product
# np.argmin(a) -> Index position of minimum value
# np.argmax(a) -> Index position of maximum value
# np.abs(a) -> Converts negative values to positive -> Absolute value (distance from zero)
# np.unique(a) -> Returns unique values only -> Removes duplicates
# np.percentile(a, 50) -> Finds percentage-based value -> 50th percentile = Median
# np.quantile(a, 0.5) -> Similar to percentile -> 0.5 = 50%
# np.ptp(a) -> Range = Max - Min
# np.any(a) -> True if at least one value is True
```

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print("Array =", a)

print("Sum =", np.sum(a))
print("Mean =", np.mean(a))
print("Median =", np.median(a))
print("Min =", np.min(a))
print("Max =", np.max(a))

print("Variance =", np.var(a))
print("Standard Deviation =", np.std(a))

print("Product =", np.prod(a))

print("Cumulative Sum =", np.cumsum(a))
print("Cumulative Product =", np.cumprod(a))

print("Index of Min =", np.argmin(a))
print("Index of Max =", np.argmax(a))

print("Absolute Value =", np.abs([-10, -20, 30]))

b = np.array([10, 20, 20, 30, 30, 40])
print("Unique Values =", np.unique(b))

print("50th Percentile =", np.percentile(a, 50))
print("0.5 Quantile =", np.quantile(a, 0.5))

print("Range (Max-Min) =", np.ptp(a))

c = np.array([False, False, True, False])
print("Any True? =", np.any(c))
print("All True? =", np.all(c))
```

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print(np.sort(a))          # Sort array
print(np.flip(a))          # Reverse array
print(np.sqrt(a))          # Square root
print(np.square(a))        # Square
print(np.power(a, 2))      # Power
print(np.log(a))           # Natural log
print(np.exp(a))           # e^x

print(np.sin(a))           # Sine
print(np.cos(a))           # Cosine
print(np.tan(a))           # Tangent

print(np.ceil([1.2, 2.5, 3.1]))   # Round up
print(np.floor([1.2, 2.5, 3.9]))  # Round down
print(np.round([1.2, 2.5, 3.7]))  # Round nearest

print(np.count_nonzero(a)) # Count non-zero values
print(np.where(a > 25))    # Index where condition is True

print(np.clip(a, 15, 45))  # Limit values between 15 and 45
```

```python
# handling missing values
import numpy as np
arr = np.array([10,20,np.nan,40])
print(arr)
#isnan
print(np.isnan(arr))

# update nan by number

new_arr = np.nan_to_num(arr,nan=100)
print(new_arr)
```

```python
# Data Structure : [restaurant_id, 2021, 2022, 2023, 2024]

import numpy as np

sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],   # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],   # Beijing Bites
    [3, 200000, 230000, 260000, 300000],   # Pizza Hub
    [4, 180000, 210000, 240000, 270000],   # Burger Point
    [5, 160000, 185000, 205000, 230000]    # Chai Point
])

# 1. Print complete dataset
print("1 --> Complete dataset :\n", sales_data)

# 2. Shape of array
print("2 --> Shape of Array :", sales_data.shape)

# 3. Number of dimensions
print("3 --> Number of Dimensions :", sales_data.ndim)

# 4. Total number of elements
print("4 --> Total number of Elements :", sales_data.size)

# 5. Restaurant IDs
print("5 --> Restaurant IDs :\n", sales_data[:, 0])

# 6. 2021 sales data
print("6 --> 2021 sales data :\n", sales_data[:, 1])

# 7. 2024 sales data
print("7 --> 2024 sales data :\n", sales_data[:, -1])

# 8. Maximum sales value
print("8 --> Max sales Value :", np.max(sales_data[:, 1:]))

# 9. Minimum sales value
print("9 --> Min sales Value :", np.min(sales_data[:, 1:]))

# 10. Total sales in 2021
print("10 --> Total sales in 2021 :", np.sum(sales_data[:, 1]))

# 11. Average sales in 2024
print("11 --> Avg sales in 2024 :", np.average(sales_data[:, -1]))

# 12. Total sales of each restaurant
print("12 --> Total sales of each restaurant :\n",
      np.sum(sales_data[:, 1:], axis=1))

# 13. Average sales of each restaurant
print("13 --> Avg sales of each restaurant :\n",
      np.average(sales_data[:, 1:], axis=1))
```
