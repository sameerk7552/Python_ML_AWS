```python
# operations on columns
import pandas as pd
data = {
    "Emp_ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Amit", "Riya", "Raj", "Sara", "John", "Neha"],
    "Department": ["IT", "HR", "Finance", "IT", "Sales", "HR"],
    "Salary": [50000, 45000, 60000, 55000, 48000, 52000],
    "Experience": [2, 3, 5, 4, 1, 3]
}
df = pd.DataFrame(data)

# drop cols
df.drop("Emp_ID",axis=1,inplace=True)
print(df)

# drop multi cols
df.drop(["Name","Department"],axis=1,inplace=True)
print(df)
```

```python
# operations on Rows
import pandas as pd
data = {
    "Emp_ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Amit", "Riya", "Raj", "Sara", "John", "Neha"],
    "Department": ["IT", "HR", "Finance", "IT", "Sales", "HR"],
    "Salary": [50000, 45000, 60000, 55000, 48000, 52000],
    "Experience": [2, 3, 5, 4, 1, 3]
}
df = pd.DataFrame(data)

# loc 1-> index of row , second -> cols name
print(df.loc[1,"Name"])

# iloc row index -> 0 and cols index -> 0
print(df.iloc[1,1])

 # get rows
print(df.loc[[0,1,2]])

# get rows using iloc
print(df.iloc[[1,2,3]])

# multi rows and cols
print(df.iloc[[0,1],[0,1]])

# slicing
print(df.iloc[0:3])
```

```python
import pandas as pd

data = {
    "Emp_ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Amit", "Riya", "Raj", "Sara", "John", "Neha"],
    "Department": ["IT", "HR", "Finance", "IT", "Sales", "HR"],
    "Salary": [50000, 45000, 60000, 55000, 48000, 52000],
    "Experience": [2, 3, 5, 4, 1, 3]
}

df = pd.DataFrame(data)

# Employees with Emp_ID > 104
print(df[df['Emp_ID'] > 104])

# Employee with Emp_ID = 106
print(df[df['Emp_ID'] == 106])

# Salary between 30000 and 50000
print(df[(df['Salary'] > 30000) & (df['Salary'] < 50000)])
```

```python
import pandas as pd

url = "https://raw.githubusercontent.com/rajendra0968jangid/Ds-Arya/main/student-data.json"

df = pd.read_json(url)
print(df)

# Filter by name
print(df[df['name'] == 'arun'])

# Physics marks greater than 80
print(df[df['physics'] > 80])

# English > 50 and Maths < 90
print(df[(df['english'] > 50) & (df['maths'] < 90)])

# Sort by English marks (ascending)
print(df.sort_values("english"))

# Sort by English marks (descending)
print(df.sort_values("english", ascending=False))

# Sort by two columns
print(df.sort_values(["english", "maths"]))
```
