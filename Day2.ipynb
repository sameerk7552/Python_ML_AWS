```python
#if else
age = 18

if age >= 18:
    print("Adult")
else:
    print("Minor")
```

```python
# if elif else
marks = 75

if marks >= 90:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
```

```python
#list
# Create a list
numbers = [1, 2, 3, 4, 5]

# Access items
print(numbers[0])  # 1

# Add an item
numbers.append(6)

# Loop through the list
for num in numbers:
    print(num)

# Check if an item exists
if 3 in numbers:
    print("Found")
```
